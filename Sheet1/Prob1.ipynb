# Problem 1.1: beware of numerical instabilities

```python
import numpy as np
import numpy.linalg as la
```

### Task 1.1.1

```python
matX = np.array([[ 1.00000, 0.00000, 0.00000],
                 [-1.00000, 0.00001, 0.00000]])

vecY = np.array( [ 0.00000, 0.00001, 0.00000] )  

vecW = la.inv(matX @ matX.T) @ matX @ vecY

print ('pen-and-paper solution:', vecW)
```

### Task 1.1.2

```python
matQ1, matR1 = la.qr(matX.T)

vecWb = la.inv(matR1) @ matQ1.T @ vecY

print ('qr solution:', vecWb)
```

### Task 1.1.3

```python
vecWc, *_ = la.lstsq(matX.T, vecY, rcond=None)

print ('lstsq solution:', vecWc)
```

Seems like lstsq() uses qr decomposition
