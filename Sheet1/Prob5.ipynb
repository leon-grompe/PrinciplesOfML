```python
import numpy as np
import numpy.linalg as la
import imageio.v3 as iio
import scipy.ndimage as img
import matplotlib.pyplot as plt
```

## Task 1.5.1

```python
# Define points
v_1 = np.array([2,8])
v_2 = np.array([8,6])
v_3 = np.array([4,2])
p   = np.array([5,5])
q   = np.array([1,3])
```

```python
# Define matrices
matV = np.column_stack((v_1, v_2, v_3))
matX = np.vstack((matV, np.array([1,1,1])))

y_p  = np.append(p, 1)
y_q  = np.append(q, 1)
```

```python
# Calculate LSQ for p
w_p = la.lstsq(matX, y_p, rcond=None)[0]

# Calculate LSQ for q
w_q = la.lstsq(matX, y_q, rcond=None)[0]
```

```python
# Check convexity constraint for w_p
print('Convexity constraint w_p: ', sum(w_p))

# Check convexity constraint for w_q
print('Convexity constraint w_q: ', sum(w_q))
```

```python
w_star_p = np.zeros((3,1))
w_star_p[0] = w_p[0]
w_star_p[1] = w_p[1]
w_star_p[2] = w_p[2]

print('w_star_p: \n', w_star_p)

# Check solution for p
result_p = matV @ w_star_p
print('result_p: \n', result_p)
```

```python
w_star_q = np.zeros((3,1))
w_star_q[0] = w_q[0]
w_star_q[1] = w_q[1]
w_star_q[2] = w_q[2]

print('w_star_q: \n', w_star_q)

# Check solution for p
result_q = matV @ w_star_q
print('result_q: \n', result_q)
```
